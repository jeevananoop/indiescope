# 06 - Les graphiques

Fonctions dans `src/charts.py` : chacune prend des données et **renvoie** une figure. Le titre dit la conclusion.
Une seule couleur neutre ; le bleu pour ce qui doit sauter aux yeux, l'orange pour le budget.

```python
import sys
sys.path.append('..')

import plotly.express as px
from src.cleaning import load_clean
from src.estimation import add_estimates
from src.analysis import niche_table, find_comparables, price_table
from src.simulator import GameConcept
from src.charts import niche_map, price_chart, revenue_distribution, releases_per_year

df = add_estimates(load_clean('../data/games_clean.csv'))
FIL_ROUGE_TAGS = ['Indie', 'Cozy']
FIL_ROUGE_BUDGET = 60000
```

## Échelle linéaire ou logarithmique ?

```python
niches = niche_table(df)
px.scatter(niches.reset_index(), x='n_games', y='median_revenue', hover_name='tag',
           title='Échelle linéaire : presque tous les points sont écrasés dans un coin').show()
```

En linéaire, quelques tags à 40 000 jeux ou à 500 000 $ écrasent tous les autres. En échelle log, chaque facteur 10 prend la même place : on garde le log.

## 1. Carte des niches

```python
fig1 = niche_map(niches, highlight=['Rogue-like', 'Cozy'])
fig1.show()
```

Lecture : chaque point est un tag. En haut à gauche, peu de concurrence et un bon revenu médian. Les deux niches du fil rouge sont en bleu.

## 2. Prix

```python
fig2 = price_chart(price_table(find_comparables(df, FIL_ROUGE_TAGS)), niche_name='Indie + Cozy')
fig2.show()
```

## 3. Mes chances face à mon budget

```python
concept = GameConcept(FIL_ROUGE_TAGS, 15, FIL_ROUGE_BUDGET)
fig3 = revenue_distribution(concept.find_comparables(df), FIL_ROUGE_BUDGET)
fig3.show()
```

## 4. La niche se remplit-elle vite ?

```python
fig4 = releases_per_year(df, FIL_ROUGE_TAGS)
fig4.show()
```

## Enregistrer les images pour le README et la note

```python
for name, fig in {'niche_map': fig1, 'price_chart': fig2,
                  'revenue_distribution': fig3, 'releases_per_year': fig4}.items():
    fig.write_image(f'../docs/images/{name}.png', width=1000, height=550)
    print('docs/images/' + name + '.png', '->', fig.layout.title.text)
```
